```
# 2nd excercise: from DNA to RNA transcript

# 1. We should define the sequence
dna_seq = "GATGGAACTTGACTACGTAAATT"

# 2. Replace T with U
#
rna_seq = dna_seq.replace("T", "U")

# 3. Print result
print(f"RNA transcript: {rna_seq}")

#N.b: whit "f" python recognizes the {} and doesn't treat it like normal text
#Just fort example print("RNA transcript: {rna_seq}") ==> out "Rna transcript:{rna_seq}"
```
