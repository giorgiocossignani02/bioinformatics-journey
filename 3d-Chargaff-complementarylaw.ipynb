```python
# 3d exercise

dna_string = "AAAACCCGGT"
complementary_dna = ""

# if cycle to replace (don't use replace for every letter to avoid errors)
# : then ; == compare

for base in dna_string:
    if base == "A":
        complementary_dna = complementary_dna + "T"
    elif base == "T":
        complementary_dna = complementary_dna + "A"
    elif base == "C":
        complementary_dna = complementary_dna + "G"
    elif base == "G":
        complementary_dna = complementary_dna + "G"

# print the string
print(f"Original: {dna_string}")
print(f"Complement: {complementary_dna}")
```
