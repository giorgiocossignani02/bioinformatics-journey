```python
# 3d exercise

dna_string = "AAAACCCGGT"
complementary_dna = "TTTTGGGCCA"

#[start:end:pace]
reverse_complement = complementary_dna[::-1]

# print the string
print(f"Original: {dna_string}")
print(f"Complement: {complementary_dna}")
print(f"Reversed: {reverse_complement}")
```
