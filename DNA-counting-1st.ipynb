```python
# First script, the easiest

# 1. Define the DNA sequence
dna_sequence = "AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC"

# 2. Using the method .count() by Python to count the number of each base type
count_A = dna_sequence.count("A")
count_C = dna_sequence.count("C")
count_G = dna_sequence.count("G")
count_T = dna_sequence.count("T")

# 3. Printing the results
print(f"Risultato: {count_A} {count_C} {count_G} {count_T}")
```
